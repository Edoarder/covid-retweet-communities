**UPLOAD DATA**

```python
#Aprile, 1-23
#Marzo: 1-31 (11 e 12 mancano)
#Febbraio: 20-29
def createList(r1, r2):
    return [item for item in range(r1, r2+1)]

month = 2 #feb = 2, march = 3, apr = 4
start_day = 25 #|| 1-23 april
last_day = 25

day_lst = createList(start_day, last_day)
```

```python
for day in day_lst:
  df = pd.read_json(f"/content/drive/MyDrive/MOTIFS PAPER/DATASETS/coronavirus_2020_0{month}_{day}.json",lines=True)
  # df = pd.read_json(f"/content/drive/MyDrive/MOTIFS PAPER/DATASETS/coronavirus_2020_0{month}_{day}.json",lines=True)

  df1 = df[['user', 'retweeted_status']]
  df1 = df[df['retweeted_status'].notna()]

  df1_userobj_raw= pd.json_normalize(df1['user'])
  df1_userobj = df1_userobj_raw[['id_str']]
  df1_A = df1_userobj.copy()
  df1_A.rename(columns={"id_str": "user_id_str"}, inplace=True)

  df1_retweetobj_raw = pd.json_normalize(df1['retweeted_status'])
  df1_retweetobj = df1_retweetobj_raw[['user.id_str']]
  df1_B = df1_retweetobj.copy()
  df1_B.rename(columns={"user.id_str": "retweet_author_id"}, inplace=True)

  df2 = pd.concat([df1_A, df1_B], axis = 1)
  df2.to_csv(f'/content/drive/MyDrive/MOTIFS PAPER/CODES/Edgelists/full_edgelist_2020_0{month}_{day}.csv', index=False)
  print(f"Month {month}, day {day} correctly processed")
```
